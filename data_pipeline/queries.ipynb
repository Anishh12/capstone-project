```python
import sqlite3
import pandas as pd
from IPython.display import display

con = sqlite3.connect("books.db")
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 60)

def run(label, sql):
    print("\n" + "=" * 78)
    print(label)
    print("-" * 78)
    print(sql.strip())
    df = pd.read_sql_query(sql, con)
    display(df)
    return df
```

```python
q1 = """
SELECT title, price_gbp, rating
FROM books
WHERE rating = 5 AND in_stock = 1
ORDER BY price_gbp DESC;
"""
df1 = run("Q1 — SELECT / WHERE (5-star, in stock)", q1)
```

```python
q2 = """
SELECT title, rating, price_inr
FROM books
ORDER BY rating DESC, price_inr DESC
LIMIT 10;
"""
df2 = run("Q2 — ORDER BY + LIMIT (top-10 by rating, then price)", q2)
```

```python
q3 = """
SELECT DISTINCT rating
FROM books
ORDER BY rating;
"""
df3 = run("Q3 — DISTINCT ratings present", q3)
```

```python
q4 = """
SELECT title, rating, price_gbp
FROM books
WHERE rating IN (4, 5)
  AND price_gbp BETWEEN 10 AND 30
ORDER BY price_gbp;
"""
df4 = run("Q4 — IN + BETWEEN (4–5 stars, £10–£30)", q4)
```

```python
q5 = """
SELECT c.category_name,
       b.title,
       b.rating,
       b.price_gbp,
       b.price_inr,
       b.in_stock
FROM books b
JOIN categories c ON b.category_id = c.category_id
WHERE b.rating >= 4
ORDER BY c.category_name, b.rating DESC, b.price_gbp DESC;
"""
df5 = run("Q5 — JOIN (4+ star books with category names)", q5)
```

```python
df_q2_sql = pd.read_sql(
    "SELECT title, rating, price_inr FROM books "
    "ORDER BY rating DESC, price_inr DESC LIMIT 10;", con
)
df_q5_sql = pd.read_sql(q5, con)     # the JOIN query from Cell 7

print("Q2 via pd.read_sql:")
display(df_q2_sql)
print("\nQ5 via pd.read_sql (the JOIN, first 10 rows):")
display(df_q5_sql.head(10))
```

```python
books_df = pd.read_sql("SELECT * FROM books;", con)
cats_df  = pd.read_sql("SELECT * FROM categories;", con)
con.close()
print("books_df:", books_df.shape, " cats_df:", cats_df.shape)
```

```python
merged = (books_df.merge(cats_df, on="category_id", how="inner")
                   .query("rating >= 4")
                   .sort_values(["category_name", "rating", "price_gbp"],
                                ascending=[True, False, False])
                   .reset_index(drop=True))

sql_join = df_q5_sql.reset_index(drop=True)
print("pd.merge result (first 10 rows):")
display(merged.head(10))
```

```python
print("=" * 78)
print("SQL JOIN vs pd.merge — side by side")
print("=" * 78)

print("\nSQL JOIN result (first 10):")
display(sql_join.head(10))

print("\npd.merge result (first 10):")
display(merged.head(10))

cols = ["category_name", "title", "rating", "price_gbp", "price_inr", "in_stock"]
equivalent = sql_join[cols].equals(merged[cols])
print(f"\nEquivalent (same columns, same order, same values): {equivalent}")
assert equivalent, "pd.merge and SQL JOIN diverged!"
print("✅ SQL JOIN and pd.merge produce identical output.")
```
